## Check NetKet

````python
import netket as nk

g = nk.graph.Chain(4)

hi = nk.hilbert.Spin(
    s=1 / 2,
    N=g.n_nodes,
)

H = nk.operator.Ising(
    hi,
    graph=g,
    h=1.0,
)

model = nk.models.RBM()

sampler = nk.sampler.MetropolisLocal(hi)

vs = nk.vqs.MCState(
    sampler,
    model,
    n_samples=128,
)

print(vs.expect(H))
````
